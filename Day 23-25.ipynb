## Step 2: MySQL se connect karo

```python
import getpass
import mysql.connector
import pandas as pd
from sqlalchemy import create_engine

HOST = "127.0.0.1"
PORT = 3306
USER = "root"
PASSWORD = getpass.getpass("kabir123 ")

conn = mysql.connector.connect(host=HOST, port=PORT, user=USER, password=PASSWORD)
cursor = conn.cursor()
print("Connected:", conn.is_connected())
```

## Step 3: Database 

```python
cursor.execute("CREATE DATABASE IF NOT EXISTS employee_db")
cursor.execute("USE employee_db")
print("Database ready")
```

## Step 4: Table 

```python
cursor.execute("DROP TABLE IF EXISTS employees")
cursor.execute("""
CREATE TABLE employees (
    emp_id     INT PRIMARY KEY AUTO_INCREMENT,
    name       VARCHAR(50) NOT NULL,
    age        INT,
    department VARCHAR(30),
    salary     INT,
    join_date  DATE
)
""")
print("Table created")
```

## Step 5: Data insert

```python
data = [
    ("Raj",    24, "Clerk",   18000, "2023-01-15"),
    ("Sita",   22, "Clerk",   15000, "2024-03-10"),
    ("Amit",   35, "Manager", 28000, "2019-07-01"),
    ("Neeraj", 40, "HR",      45000, "2017-05-20"),
    ("Kavita", 28, "IT",      32000, "2021-09-12"),
    ("Deepak", 23, "IT",      19000, "2024-01-08"),
    ("Suman",  45, "Manager", 60000, "2015-11-30"),
]

cursor.executemany(
    "INSERT INTO employees (name, age, department, salary, join_date) VALUES (%s,%s,%s,%s,%s)",
    data
)
conn.commit()
print(cursor.rowcount, "rows inserted")
```

## Step 6: Data 

```python
engine = create_engine(f"mysql+pymysql://{USER}:{PASSWORD}@{HOST}:{PORT}/employee_db")
df = pd.read_sql("SELECT * FROM employees", engine)
df
```

## Step 7: Queries 

```python
pd.read_sql("SELECT * FROM employees WHERE salary > 30000 OR department = 'Manager'", engine)
```

```python
pd.read_sql("SELECT department, AVG(salary) AS avg_salary FROM employees GROUP BY department", engine)
```

```python
pd.read_sql("SELECT * FROM employees WHERE salary < 20000 AND age < 25", engine)
```

## Step 8: 

```python

cursor.execute("INSERT INTO employees (name, age, department, salary, join_date) VALUES ('Mohit', 26, 'IT', 27000, '2025-02-01')")

cursor.execute("UPDATE employees SET salary = 20000 WHERE name = 'Raj'")

cursor.execute("DELETE FROM employees WHERE name = 'Sita'")
conn.commit()
pd.read_sql("SELECT * FROM employees", engine)
```

## Step 9: Connection close

```python
cursor.close()
conn.close()
engine.dispose()
print("Closed")
```
